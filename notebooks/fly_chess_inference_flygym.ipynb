# Fly Chess with optional 3D fly

This notebook runs the regular inference interface and adds an optional FlyGym view. Leave **Show 3D fly** off for faster games; turn it on to render a short 3D simulation after each fly move.


```python
from pathlib import Path
import os
import subprocess
import sys

IN_COLAB = "COLAB_RELEASE_TAG" in os.environ or "COLAB_GPU" in os.environ
PROJECT_DIR = Path("/content/making-fly-play-chess") if IN_COLAB else Path.cwd()

if IN_COLAB and not PROJECT_DIR.exists():
    subprocess.check_call([
        "git", "clone",
        "https://github.com/mncrftfrcnm/making-fly-play-chess.git",
        str(PROJECT_DIR),
    ])

if not (PROJECT_DIR / "fly_chess_model.joblib").is_file():
    raise FileNotFoundError(f"Could not find the project files in {PROJECT_DIR}")

print(f"Using {PROJECT_DIR}")
```

```python
subprocess.check_call([
    sys.executable, "-m", "pip", "install", "-q",
    "-r", str(PROJECT_DIR / "requirements.txt"),
])

if sys.version_info >= (3, 12):
    if IN_COLAB:
        os.environ.setdefault("MUJOCO_GL", "egl")
        os.environ.setdefault("PYOPENGL_PLATFORM", "egl")
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "flygym==2.1.0"
    ])
else:
    print("FlyGym 2.1.0 requires Python 3.12+. The normal interface still works with the 3D view off.")
```

## Start the app

The 3D view is disabled by default because MuJoCo simulation and video rendering make each fly move slower.


```python
import runpy
import sys

sys.path.insert(0, str(PROJECT_DIR / "scripts"))

runpy.run_path(
    str(PROJECT_DIR / "scripts" / "fly_chess_inference_flygym.py"),
    run_name="__main__",
)
```
